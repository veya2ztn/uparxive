```python
import json
import os
from tqdm.auto import tqdm
def get_location(data):
    url = data.get('landing_page_url', None)
    key = None
    if url is not None:
        key = os.path.split(url)[-1]
    if 'source' not in data or data['source'] is None:
        return {'journal':  None,
                'issn_l':   None,
                'issn':     None,
                'publisher':None,
                'url': url,
                'key': key
               }
    return {'journal':  data['source'].get('display_name', None),
            'issn_l':   data['source'].get('issn_l', None),
            'issn':     data['source'].get('issn', None),
            'publisher':data['source'].get('publisher', None),
            'url': url,
            'key': key
           }


json_pack_file = "works/updated_date=2023-10-29/part_000"

with open('openalex.jsonl', 'w') as f:
    new_jsonl      = []
    paper_identity = []
    with open(json_pack_file, 'r') as f:
        for line in tqdm(f):
            data = json.loads(line.strip())
            paper_identity.append(data['ids'])
            new_data = {
                'openalex_id': data.get('id',None),
                'doi': data.get('doi', None),
                'title': data.get('title', None),
                'publication_date': data.get('publication_date', None),
                'language': data.get('language', None),
                'authors': [t['author']['display_name'] for t in data['authorships']],
                'location': [get_location(t) for t in data['locations']],
                'referenced_works': data['referenced_works'],
            }
            new_jsonl.append(new_data)

    for line in new_jsonl:
        f.write(json.dumps(line)+'\n')
with open('openalex.jsonl', 'w') as f:
    for line in new_jsonl:
        f.write(json.dumps(line)+'\n')
```
