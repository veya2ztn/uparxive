````python
def get_puyu_judge(sentences):
    messages = [{"role": "user", "content": f"""
    Your job is to identify a sentense is valid for reference finding. You only need answer yes or no. 
    ```
    Below are some examples that valid for reference finding: 
    G. Bressi, G. Carugno, R. Onofrio and G. Ruoso, Phys. Rev. Lett. 88 041804 (2002)
    R.S. Decca, D. López, E. Fischbach and D. E. Krause, ibid. 91 050402 (2003).
    N.G. Van Kampen, B.R.A. Nijboer and K. Schram, Phys. Lett. 26A 307 (1968)
    J. Heinrichs, Phys. Rev. B11 3625 (1975)
    C. Genet, F. Intravaia, A. Lambrecht and S. Reynaud, Ann. Fond. L. de Broglie 29 311 (2004)
    arXiv:quant-ph/0302072.
    ```.
    Notice, there are also some cases that won't valid for reference. For example, if there is only one author but without the information of publisher like
    ```
    S. Strauß et al. , Submitted for publication.
    Ch. J. Bordé in reference 4, p 257
    H.-D. Gräf, Technische Universität Darmstadt, private communication.
    ```
    Now, please judge follow sentense:{sentence}.

    Notice,  You only need answer yes or no."""} for sentence in sentences]

    results = []
    for item in api_client.chat_completions_v1(model=model_name, messages=messages):
        results.append(item)
    return results
````

````python
def is_email(sentence):
    return 'mail' in sentence.lower()
def is_web(sentence):
    sentence = sentence.lower()
    return 'www'in sentence or 'http 'in sentence or 'html' in sentence
def is_address(sentence):
    return 'address'in sentence.lower()
def contains_arxiv_id(text):
    # Regular expression pattern for matching ArXiv IDs
    # It matches "arXiv:" or "arXiv" followed by optional space(s) and the ID pattern
    arxiv_pattern = r'arxiv:? *\s*(\d{4})\.(\d{3,4})(v\d+)?'
    
    # Search for the pattern in the text
    match = re.search(arxiv_pattern, text)
    
    # Return True if a match is found, False otherwise
    return match
def is_citation_string(sentence):
    # Check for citation pattis_citation_stringerns and a four-digit year
    if is_email(sentence):return False
    if is_web(sentence):return False
    if is_address(sentence):return False
    if has_four_digit_year(sentence) or has_citation_pattern(sentence):
        return True
    if has_slash_followed_by_digits(sentence):
        return True
    if len(find_citation_names(sentence))>1:
        return True
    if bool(contains_arxiv_id(sentence.lower().strip('.'))):
        return True
    return False
````

````python
import re

def find_citation_names(text):
    # This pattern matches a single initial followed by a full last name
    # It can also match a pattern of multiple initials representing first and middle names, followed by a full last name
    # It supports compound last names with hyphens and Unicode characters
    # It also handles cases where there is no space after a comma
    name_pattern = re.compile(r"""
        \b                                      # Word boundary to ensure we're matching full words
        (?:[A-Z]\.?\s*)+                        # One or more initials with optional periods followed by optional spaces
        [A-Z][\w'-]+                            # Matches the last name starting with an uppercase letter followed by letters, which can include Unicode characters, apostrophes or hyphens
        (?:\s?-?\s?[A-Z][\w'-]+)*              # Matches compound last names with optional spaces and hyphens
        \b                                      # Word boundary to ensure we're matching full words
    """, re.VERBOSE | re.UNICODE)

    # Find all matches in the text
    return name_pattern.findall(text)

# Example usage:
citation_text = "In the study by J. H. Doe and A. B. Chain (2023), the authors concluded that their results were significant. M. O'Connor (2020) offered a different perspective."

names = find_citation_names(citation_text)
print("Names found in citation:")
for name in names:
    print(name)
````

````python

import argparse
#--------------------------------------------------#
def check_is_args(args):
    if dataclasses.is_dataclass(args):return True
    if isinstance(args, argparse.Namespace):return True
    return False
def get_print_namespace_tree(namespace, indent=0):
    result = ""
    namespace = vars(namespace) if check_is_args(namespace) else namespace
    for key, value in namespace.items():
        line = ' ' * indent
        if isinstance(value, dict) or check_is_args(value):
            line += key + "\n"
            line += get_print_namespace_tree(value, indent + 4)
        else:
            line += f"{key:30s} ---> {value}\n"
        result += line
    return result 

def print_namespace_tree(namespace):
    print(get_print_namespace_tree(namespace))
````

````python
from typing import List
from dataclasses import dataclass
import dataclasses
@dataclass
class Reference:
    unique_id: str
    title: str
    author: List[str]
    journal: str
    journal_volume: int
    journal_page: int
    year: int
    publisher: str
    content: str
    
````

````python

def obtain_reference_list(tmp_xml_path):
    reference_string_path = tmp_xml_path.replace('reference.grobid.tei.xml','reference.txt')
    with open(reference_string_path,'r') as f:
        string_list = [line.strip() for line in f ]
    parser = etree.XMLParser(remove_comments=True)
    with open(tmp_xml_path) as f:
        tree = etree.parse(f, parser)  # get tree of XML hierarchy
    root=tree
    # Define the namespaces used in the XML data
    namespaces = {
        'tei': "http://www.tei-c.org/ns/1.0",
    }
    biblStructs = root.findall('.//tei:biblStruct', namespaces)
    assert len(biblStructs) == len(string_list), f"num_of_structured_ref:{len(biblStructs)} num_of_ref_string:{len(string_list)}"

    reference_list=[]
    # Initialize an empty dictionary to store the extracted information
    bibliography = []

    # Iterate through each 'biblStruct' element in the XML
    for biblStruct,citation_string in zip(biblStructs,string_list):
        arxiv_match = contains_arxiv_id(citation_string.lower().strip('.'))
        unique_id = arxiv_match.group(0) if arxiv_match else None
        # Extract the ID of the biblStruct
        bib_id = biblStruct.get('{http://www.w3.org/XML/1998/namespace}id')
        
        # Initialize a dictionary to store the current biblStruct data
        bib_info = {
            'title': None,
            'author': [],
            'imprint': {},
            'note': None,
            'journal': None,
            'journal_volume': None,
            'publisher':None,
            'content': citation_string,
            'unique_id':unique_id
        }

        # Extract analytic (article-level) information, if any
        for analytic in biblStruct.findall('tei:analytic', namespaces):
            for title in analytic.findall('tei:title', namespaces):
                bib_info['title']=(title.text or None)
            for author in analytic.findall('.//tei:author/tei:persName', namespaces):
                forename = ".".join(name.text.strip() for name in author.findall('.//tei:forename', namespaces))
                surname  = ".".join(name.text.strip() for name in author.findall('.//tei:surname', namespaces))
                author_name = f"{forename} {surname}"
                bib_info['author'].append(author_name)

        # Extract monographic (book-level) information, if any
        for monogr in biblStruct.findall('tei:monogr', namespaces):
            for title in monogr.findall('tei:title', namespaces):
                bib_info['journal']=(title.text or None)
            for imprint in monogr.findall('tei:imprint', namespaces):
                for item in imprint:
                    if item.tag.endswith('biblScope'):
                        bib_info['imprint'][item.get('unit')] = item.get('from', item.text)
                    elif item.tag.endswith('date'):
                        bib_info['imprint']['date'] = item.get('when')
                    elif item.tag.endswith('publisher'):
                        bib_info['imprint']['publisher'] = item.text

        # Extract notes, if any
        for note in biblStruct.findall('tei:note', namespaces):
            bib_info['note']=(note.text or None)
        
        
            
    #     reference_list.append(Reference(
    #         unique_id = unique_id,
    #             title = bib_info['titles'][0] if len(bib_info['titles'])>0 else None,
    #             author= bib_info['authors'][0] if len(bib_info['authors'])>0 else None,
    #             journal=bib_info['journal'][0],
    #             journal_volume=bib_info['imprint'].get('volume',None),
    #             journal_page=bib_info['imprint'].get('page',None),
    #             year = bib_info['imprint'].get('year',None),
    #             publisher = bib_info['imprint'].get('publisher',None),
    #         content=citation_string,
            
    #     ))
        # Add the current biblStruct to the bibliography dictionary
        bibliography.append(bib_info)

    return bibliography
````

````python
def get_anystyle_ref_query_dict_v21(info):
    search_should = []
    flags = defaultdict(int)
    if 'author' in info and info['author']:
        for author_i in info['author'][:3]:
            if isinstance(author_i,dict):
                name = HumanName(" ".join([author_i.get("given", ""), author_i.get("family", "")]).strip(),initials_format="{first} {middle}")
            elif isinstance(author_i, str):
                name = HumanName(full_name=author_i)
            name = name.initials() + ' ' + name.last
            search_should.append({
                "multi_match": {"query": "{}".format(name),"fields": ["author.0", "author.1", "author.2"],"type": "best_fields"}
            })
            flags['author'] += 1
    volume = (info.get('volume',None) or info.get('journal_volume',None))
    if volume:
        search_should.append({
            "multi_match": {"query": strfy(volume),"fields": ["jvol"],"type": "best_fields"}
        })
        flags['volume'] = 1
    
    page = (info.get('page',None) or info.get('journal_page',None))
    if page:
        search_should.append({
            "multi_match": {"query": strfy(page),"fields": ["jpage"],"type": "best_fields"}
        })
        flags['page'] = 1
    
    title = (info.get('title',None) or info.get('title',None))
    if title:
        search_should.append({
            "multi_match": {"query": strfy(title),"fields": ["title" ],"type": "best_fields"}
        })
        flags['title'] = 1
                      
    journal = (info.get('container-title',None) or info.get('journal',None) or info.get('publisher',None))
    if journal:
        search_should.append({
            "multi_match": {"query": strfy(journal),"fields": ["jname"],"type": "best_fields"}
        })
        flags['vene'] = 1
    
    
    year = None
    if not year and 'issued' in info and info['issued'] and 'date-parts' in info['issued']:
        date_part = info['issued']['date-parts']
        while isinstance(date_part, list):
            date_part = date_part[0] 
        year = date_part
        
    if not year:
        year = info.get('year', None)
    
    if year:
        search_should.append({
            "multi_match": {"query": strfy(year),"fields": ["year"],"type": "best_fields"}
        })
        flags['year'] = 1

    
    ### form the query 
    query_dict = {
        "bool": {
            "should": search_should
        }
    }
    return query_dict
````

````python
def similarity_structured(self, candidate, ref):
    # weights for generalized jaccard similarity
    cand_set = {}
    str_set = {}

    # weights of volume
    if ref.get('volume', ''):
        self.update_weights_one('volume', candidate.get('volume', ''),
                                ref.get('volume', ''), cand_set, str_set)

    # weights for year
    if ref.get('year', ''):
        year  = candidate.get('year', '')
        self.update_weights_one('year', year, ref.get('year', ''),
                                cand_set, str_set)
        if 'year' in str_set and str_set['year'] < 1 and year \
                and ref.get('year', ''):
            try:
                year1 = int(year)
                year2 = int(ref.get('year', ''))
                if year1 + 1 == year2 or year2 + 1 == year1:
                    str_set['year'] = 0.5
            except ValueError:
                pass

    # weights for pages
    if ref.get('first-page', ''):
        self.update_weights_one('page', candidate.get('page', ''),
                                ref.get('first-page', ''),
                                cand_set, str_set)

    # weights for title
    if ref.get('article-title', ''):
        a = self.get_cand_norm(candidate, 'title')
        b = self.get_ref_norm(ref, 'article-title')
        cand_set['title'] = 1
        str_set['title'] = fuzz.ratio(a, b) / 100

    # weights for container-title
    if ref.get('journal-title', ''):
        a = self.get_cand_norm(candidate, 'container-title')
        b = self.get_ref_norm(ref, 'journal-title')
        cand_set['ctitle'] = 1
        str_set['ctitle'] = fuzz.ratio(a, b) / 100

    # weights for volume-title
    if ref.get('volume-title', ''):
        a = self.get_cand_norm(candidate, 'title')
        b = self.get_ref_norm(ref, 'volume-title')
        title_ratio = fuzz.ratio(a, b)
        a = self.get_cand_norm(candidate, 'container-title')
        ctitle_ratio = fuzz.ratio(a, b)
        cand_set['vtitle'] = 1
        str_set['vtitle'] = max(title_ratio, ctitle_ratio) / 100

    # weights for author
    if ref.get('author', ''):
        a = ''
        if 'author' in candidate and candidate['author'] \
                and 'family' in candidate['author'][0]:
            a = unidecode.unidecode(candidate['author'][0]['family'])
            a = a.lower().strip()
        b = self.get_ref_norm(ref, 'author')
        ratio_author = fuzz.partial_ratio(a, b) if ' ' in b \
            else fuzz.ratio(a, b)
        a = ''
        if 'editor' in candidate and candidate['editor'] \
                and 'family' in candidate['editor'][0]:
            a = unidecode.unidecode(candidate['editor'][0]['family'])
            a = a.lower().strip()
        ratio_editor = fuzz.partial_ratio(a, b) if ' ' in b \
            else fuzz.ratio(a, b)
        cand_set['author'] = 1
        str_set['author'] = max(ratio_author, ratio_editor) / 100

    support = 0
    for k, v in str_set.items():
        if k == 'title' and v > 0.7:
            support = support + 1
        if k == 'ctitle' and v > 0.7:
            support = support + 1
        if k == 'vtitle' and v > 0.7:
            support = support + 1
        if k == 'author' and v > 0.7:
            support = support + 1
    if support < 1:
        return 0
    support = 0
    for k, v in str_set.items():
        if k == 'year' and v > 0 and cand_set[k] > 0:
            support = support + 1
        if k == 'volume' and v == 1 and cand_set[k] > 0:
            support = support + 1
        if k == 'title' and v > 0.7:
            support = support + 1
        if k == 'ctitle' and v > 0.7:
            support = support + 1
        if k == 'vtitle' and v > 0.7:
            support = support + 1
        if k == 'author' and v > 0.7:
            support = support + 1
        if k == 'page' and v == 1 and cand_set[k] > 0:
            support = support + 1
    if support < 3:
        return 0
    if candidate.get('type') == 'book-chapter' and 'first-page' not in ref:
        return 0
    if candidate.get('type') == 'journal-issue':
        return 0

    # generalized Jaccard similarity
    num = sum([min(cand_set[n], str_set[n]) for n in cand_set.keys()])
    den = sum([max(cand_set[n], str_set[n]) for n in cand_set.keys()])
    if den == 0:
        return 1
    return num/den
````
